```python
import pandas as pd

df = pd.read_csv('../data/car_sales_cleaned1.csv')
df
```

```python
df = df.drop_duplicates()  # removing duplicates
df.shape
```

```python
df = df.dropna()  # removing rows with missing values
df.shape
```

```python
import joblib
import converter

model = joblib.load('../models/car_gender_model.pkl')
mapper = converter.columns_mapper(['Model', 'Transmission', 'Color'], df)
prediction = model.predict([[mapper['Model']['Escort'], mapper['Transmission']['Auto'], mapper['Color']['Pale White']]])
prediction
```

```python
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
import joblib
import converter
model = DecisionTreeClassifier()

mapper = converter.columns_mapper(['Gender','Model', 'Color'], df)
df_converted = converter.convert_x(df, mapper)
X = df_converted[['Color','Gender']]
y = df_converted[['Model']]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
model.fit(X_train, y_train)
prediction = model.predict(X_test)
score = accuracy_score(y_test, prediction)

print(score)
#joblib.dump(model, '../models/car_predictor_model.pkl')
```
